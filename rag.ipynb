```python
import os 
from dotenv import load_dotenv
load_dotenv()



from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import JinaEmbeddings
```

```python

groq_api_key = os.getenv("GROQ_API_KEY")
jina_api_key = os.getenv("JINA_API_kEY")
```

```python


path = os.path.join("data","policy.txt")
print(path)

```

```python
# data ingestion

loader = TextLoader(path,encoding="utf-8")

documents = loader.load()

```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
split_documents = splitter.split_documents(documents)
type(split_documents)
```

```python
print(split_documents[0])
```

```python
# now we splitted the data into chunks now its time to embed the data
```

```python
jina_embeddings = JinaEmbeddings(model_name="jina-embeddings-v2-base-en")
```

```python
# store vector in vector store which is local way to store the data
```

```python
from langchain_community.vectorstores import FAISS

vector_store = FAISS.from_documents(split_documents,jina_embeddings)
print(vector_store.index.ntotal)
```

```python
vector_store
```

```python
# test the vector store
```

```python
test_query = "what is the policy for the employee to restrict sexual harassment" 
```

```python
top_matches = vector_store.similarity_search(test_query,k=2)
for match in top_matches:
    print(match.page_content)
```
